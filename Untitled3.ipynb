```python
# Assuming 'Date' is the name of the column containing dates:
correlation = gold.drop(columns=['Date']).corr()
gold['Date'] = pd.to_datetime(gold['Date'])
gold['DayOfYear'] = gold['Date'].dt.dayofyear
correlation = gold.drop(columns=['Date']).corr() # Exclude original Date column
```

```python
sns.distplot(gold['GLD'],color='green')
```

```python
x=gold.drop(['Date','GLD'],axis=1)
Y= gold['GLD']
print(x)
```

```python
print(Y)
```

```python
x_train, x_test, Y_train, Y_test= train_test_split(x,Y,test_size=0.2,random_state=2)
regressor =RandomForestRegressor(n_estimators=100)
```

```python
#training the model
regressor.fit(x_train,Y_train)
```

```python
#prediction on test data
test_Data_prediction= regressor.predict(x_test)
print(test_Data_prediction)
```

```python
# R square error
error_score=metrics.r2_score(Y_test, test_Data_prediction)
print("R square error:",error_score )
```

```python
plt.plot(Y_test, color='blue', label='Actual value') # Changed 'lable' to 'label'
plt.plot(test_Data_prediction, color='green', label='predicted value') # Changed 'lable' to 'label'
plt.title('Actual price vs predicted price')
plt.xlabel('Number of values') # Changed 'xlable' to 'xlabel'
plt.ylabel('GLD price') # Changed 'ylable' to 'ylabel'
plt.legend()
plt.show()
```
